```python
"""
Docstring for spark.manishgithubrepo.notebooks.leetcode.rank_scores_q_178.ipynb
Table: Scores

+-------------+---------+
| Column Name | Type    |
+-------------+---------+
| id          | int     |
| score       | decimal |
+-------------+---------+
id is the primary key (column with unique values) for this table.
Each row of this table contains the score of a game. Score is a floating point value with two decimal places.
 

Write a solution to find the rank of the scores. The ranking should be calculated according to the following rules:

The scores should be ranked from the highest to the lowest.
If there is a tie between two scores, both should have the same ranking.
After a tie, the next ranking number should be the next consecutive integer value. In other words, there should be no holes between ranks.
Return the result table ordered by score in descending order.

The result format is in the following example.

 

Example 1:

Input: 
Scores table:
+----+-------+
| id | score |
+----+-------+
| 1  | 3.50  |
| 2  | 3.65  |
| 3  | 4.00  |
| 4  | 3.85  |
| 5  | 4.00  |
| 6  | 3.65  |
+----+-------+
Output: 
+-------+------+
| score | rank |
+-------+------+
| 4.00  | 1    |
| 4.00  | 1    |
| 3.85  | 2    |
| 3.65  | 3    |
| 3.65  | 3    |
| 3.50  | 4    |
+-------+------+

"""
```

```python
from pyspark.sql import SparkSession
from pyspark.sql.types import StructType, StructField, IntegerType, DecimalType
from decimal import Decimal
spark = SparkSession.builder.getOrCreate()

scores_schema = StructType([
    StructField("id", IntegerType(), False),
    StructField("score", DecimalType(4, 2), True),
])

scores_data = [
    (1, Decimal("3.50")),
    (2, Decimal("3.65")),
    (3, Decimal("4.00")),
    (4, Decimal("3.85")),
    (5, Decimal("4.00")),
    (6, Decimal("3.65")),
]

scores_df = spark.createDataFrame(scores_data, schema=scores_schema)
```

```python
scores_df.show()
```

```python
from pyspark.sql.functions import *
from pyspark.sql.window import Window
```

```python
window_specs=Window.orderBy(col("score").desc())
```

```python
scores_df.withColumn("ranks",dense_rank().over(window_specs)).select(col("score"),col("ranks")).show()
```
