```python
from pyspark.sql import SparkSession

spark = (
    SparkSession.builder
    .appName("flatmap")
    .master("local[*]")
    # 🔑 CRITICAL FIXES
    .config("spark.driver.bindAddress", "0.0.0.0")
    .config("spark.driver.host", "127.0.0.1")
    .config("spark.ui.host", "0.0.0.0")
    .config("spark.ui.port", "4040")
    .getOrCreate()
)


```

```python
from pyspark.sql.types import StructType,StructField,StringType
schema=StructType(
    [
        StructField("sentence",StringType(),True)
    ]
)
sentences = [
    "spark is fast",
    "spark is powerful"
]

data=spark.createDataFrame([[i,] for i in sentences],schema)
```

```python
from pyspark.sql.functions import col
```

```python
def split_fn(x):
    splitted=x['sentence'].split()
    return splitted
splitted_data=data.select(col('sentence')).rdd.flatMap(split_fn).collect()
```

```python
data.select(col("sentence")).rdd.map(lambda x:x['sentence'].split()).collect()
```

```python
data.select(col("sentence")).rdd.flatMap(lambda x:x['sentence'].split()).collect()
```
